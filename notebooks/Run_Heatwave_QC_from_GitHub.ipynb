# Heatwave QC — run from GitHub
This notebook runs the independent public-data sensitivity benchmark. **Synthetic flags are NOT NOAA's real QC flags.** Do not publish demo figures as findings.


```python
REPO_URL = 'https://github.com/priambudihudaf/projects-heatwave-qc-policy-benchmark.git'
REPO_DIR = '/content/projects-heatwave-qc-policy-benchmark'
PROJECT_SUBDIR = '.'
STATIONS = 12
REPS = 30
```

```python
import subprocess, sys, os
from pathlib import Path
repo = Path(REPO_DIR)
if not repo.exists():
    subprocess.run(['git','clone','--depth','1',REPO_URL,str(repo)], check=True)
else:
    subprocess.run(['git','-C',str(repo),'pull','--ff-only','origin','main'], check=True)
project = repo / PROJECT_SUBDIR
os.chdir(project)
subprocess.run([sys.executable,'-m','pip','install','-q','-r','requirements.txt'], check=True)
print('Project:', project)
print('Commit:', subprocess.check_output(['git','rev-parse','HEAD'], cwd=repo, text=True).strip())
```

```python
subprocess.run([sys.executable,'-m','unittest','discover','-s','tests','-v'],check=True)
subprocess.run([sys.executable,'run.py','--mode','demo','--stations','4','--reps','3'],check=True)
print('Demo is entirely synthetic and NOT a scientific result.')
```

```python
# Real NOAA Daily01 data, but simulated QC flags/errors. Sources are public NOAA.
subprocess.run([sys.executable,'run.py','--mode','real','--stations',str(STATIONS),'--reps',str(REPS)],check=True)
print('Pilot result directory:', project / 'outputs' / 'NOAA_PILOT')
```
