# Heatwave QC Pilot v2 — controlled review prioritization
NOAA Daily01 data are quality-screened reference values. Synthetic error injection and simulated QC flags are NOT observed NOAA QC actions. The simulated reviewer is an *oracle* that perfectly restores reference values.


```python
REPO = 'https://github.com/priambudihudaf/projects-heatwave-qc-policy-benchmark.git'
BRANCH = 'research/event-impact-aware-qc-v2'
LOCAL = '/content/heatwave-qc-v2'
STATIONS = 12
REPS = 12
```

```python
from pathlib import Path
import subprocess, sys, os
repo = Path(LOCAL)
if not repo.exists():
    subprocess.run(['git','clone','--single-branch','--branch',BRANCH,'--depth','1',REPO,str(repo)],check=True)
else:
    subprocess.run(['git','-C',str(repo),'pull','--ff-only','origin',BRANCH],check=True)
os.chdir(repo)
subprocess.run([sys.executable,'-m','pip','install','-q','-r','requirements.txt'],check=True)
print('Git commit:',subprocess.check_output(['git','rev-parse','HEAD'],cwd=repo,text=True).strip())
```

```python
# Offline tests and entirely synthetic software smoke test
subprocess.run([sys.executable,'-m','unittest','discover','-s','tests','-v'],check=True)
subprocess.run([sys.executable,'run_v2.py','--mode','demo','--stations','3','--reps','2'],check=True)
```

```python
# Real PUBLIC NOAA Daily01, but artificial errors/flags and oracle review
subprocess.run([sys.executable,'run_v2.py','--mode','real','--stations',str(STATIONS),'--reps',str(REPS)],check=True)
```
